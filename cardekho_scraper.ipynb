```python
import mysql.connector

# Connection details
host = '127.0.0.1'  # or 'localhost'
database = 'student'  # Replace with your database name
user = 'root'  # Replace with your MySQL username
password = 'root'  # Replace with your MySQL password

try:
    # Establish the connection
    conn = mysql.connector.connect(
        host=host,
        database=database,
        user=user,
        password=password,
        autocommit=True
    )
   
    if conn.is_connected():
        print(f"✅ Successfully connected to MySQL database: {database}")

    # Create a cursor object
    cursor = conn.cursor()
   
    df = pd.read_csv('used_cars_8_to_10_lakh.csv')
    
    # Drop rows where 'Name' is missing
    df = df.dropna(subset=['Name'])
    # Replace NaN values with None (MySQL-compatible NULL)
    df = df.where(pd.notna(df), None)

    
    # Define the SQL query to insert data
    sql_insert_query = """
        INSERT INTO used_cars (name, price, kilometers_driven, fuel_type, transmission_type)
        VALUES (%s, %s, %s, %s, %s)
    """
    
    # Iterate over the rows in the DataFrame and insert them into the database
    for index, row in df.iterrows():
        car_data = (row['Name'], row['Price'], row['Kilometers_Driven'], row['Fuel_Type'], row['Transmission_Type'])
        print(car_data)
        cursor.execute(sql_insert_query, car_data)
        
    conn.commit;
    print("Data inserted successfully")
except mysql.connector.Error as e:
    print(f"❌ Error connecting to MySQL: {e}")

finally:
    if 'conn' in locals() and conn.is_connected():        
        cursor.close()
        conn.close()
        print("🔌 Connection closed!")
```
